```python
#https://docs.google.com/spreadsheets/d/1IthXGpmNuWNLft8RJFWmum8qHukppsrlKJraX2v6-hU/edit?usp=sharing


import pandas as pd
import requests
import time

endpoint_planilha ="https://docs.google.com/spreadsheets/d/1IthXGpmNuWNLft8RJFWmum8qHukppsrlKJraX2v6-hU/export?format=csv"
endoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print("---------- Dados Coletados ----------")
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
#iterar cada resposta do formulario
for indece, linha in dados_planilha.iterrows():
  nome_local = linha["Nome do local"]
  lat = linha["Latitude"]
  lon =  linha["Longitude"]

  parametros = {
      "latitude":lat,
      "longitude":lon,
      "current": "temperature_2m"
  }

  resposta = requests.get(endoint_api, params=parametros).json()

  display(resposta)
  print("\n\n")
```
